#📌 Extracción

```python
import pandas as pd
```

```python
datos = pd.read_json('TelecomX_Data.json')
```

```python
datos.head()
```

#🔧 Transformación

```python
datos.info()
```

```python
datos['customer'].head()
```

```python
df_normalizado = pd.json_normalize(datos['account'])
df_normalizado.head()
```

```python
df_datos = pd.json_normalize(datos.to_dict(orient='records'))
df_datos.head()
```

```python
df_datos = df_datos.drop(labels=['customerID', 'account.Charges.Total'], axis=1, inplace=False)
```

```python
df_datos
```

```python
df_datos.info()
```

```python
df_datos.duplicated().sum()
```

```python
df_datos = df_datos.drop_duplicates()
```

```python
df_datos.duplicated().sum()
```

```python
df_datos.isnull().sum()
```

```python
import numpy as np

df_datos['customer.SeniorCitizen'] = df_datos['customer.SeniorCitizen'].astype(np.int64)
```

```python
df_datos['customer.tenure'] = df_datos['customer.tenure'].astype(np.int64)
```

```python
df_datos
```

```python
df_datos['account.Charges.Monthly'] = df_datos['account.Charges.Monthly'].astype(np.float64)
df_datos.info()
```

```python
df_datos['account.Contract']
```

```python
df_datos['cuentas_diarias'] = (df_datos['account.Charges.Monthly'] * 12) / 365
df_datos
```

```python
df_datos.columns = df_datos.columns.str.lower().str.replace('.', '_')
df_datos
```

```python
cols_bol = ['churn', 'customer_partner', 'customer_dependents', 'phone_phoneservice', 'phone_multiplelines', 'internet_onlinesecurity', 'internet_onlinebackup', 'internet_deviceprotection', 'internet_techsupport', 'internet_streamingtv', 'internet_streamingmovies', 'account_paperlessbilling']
df_datos[cols_bol] = df_datos[cols_bol].replace({"Yes":1, "No":0, "No phone service":0, "No internet service":0})
df_datos.info()
```

```python
df_datos['churn'] = df_datos['churn'].replace({"":None})
df_datos = df_datos.dropna()
df_datos.groupby('churn').count()
```

```python
df_datos['churn'] = df_datos['churn'].astype(np.int64)
df_datos.info()
```

```python
df_datos['customer_gender'].value_counts()
```

```python
df_datos['internet_internetservice'].value_counts()
```

```python
df_datos['account_contract'].value_counts()
```

```python
df_datos['account_paymentmethod'].value_counts()
```

```python
df_datos.sample(10)
```

```python
df_datos['account_contract'] = df_datos['account_contract'].replace('[^a-zA-Z0-9]', ' ', regex=True)
df_datos['account_paymentmethod'] = df_datos['account_paymentmethod'].replace('[^a-zA-Z0-9]', ' ', regex=True)
```

```python
df_datos.sample(10)
```

```python
for col in ['customer_gender', 'internet_internetservice', 'account_contract', 'account_paymentmethod']:
    df_datos[col] = df_datos[col].astype(str).str.strip()
df_datos.info()
```

```python
# Convert specified object columns to pandas StringDtype
for col in ['customer_gender', 'internet_internetservice', 'account_contract', 'account_paymentmethod']:
    df_datos[col] = df_datos[col].astype(pd.StringDtype())

# Display the info again to see the updated dtypes
df_datos.info()
```

#📊 Carga y análisis

```python
df_datos.describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10,4))

# Conteo
counts = df_datos['churn'].value_counts()

# Gráfico de barras
ax[0].bar(['No Churn','Churn'], counts)
ax[0].set_title('Conteo')
ax[0].set_ylabel('Clientes')

# Gráfico de pastel
ax[1].pie(counts, labels=['No Churn','Churn'], autopct='%1.1f%%')
ax[1].set_title('Proporción')

plt.tight_layout()
plt.show()
```

```python
pd.crosstab(df_datos['customer_gender'], df_datos['churn'], normalize='index')
```

```python
import seaborn as sns

sns.countplot(x='customer_gender', hue='churn', data=df_datos)

plt.title("Churn por género")
plt.legend(['No Churn','Churn'])
plt.show()
```

```python
pd.crosstab(df_datos['internet_internetservice'], df_datos['churn'], normalize='index')
```

```python
sns.countplot(x='internet_internetservice', hue='churn', data=df_datos)

plt.title("Churn por servicio de internet")
plt.legend(['No Churn','Churn'])
plt.show()
```

```python
sns.countplot(x='account_contract', hue='churn', data=df_datos)

plt.title("Churn por tipo de contrato")
plt.legend(['No Churn','Churn'])
plt.show()
```

```python
df_datos.describe()
```

```python
num_cols = ['customer_tenure','account_charges_monthly','cuentas_diarias']

for col in num_cols:
    plt.figure(figsize=(10,5))
    sns.histplot(hue='churn', x=col, data=df_datos, bins=30, kde=True)
    plt.title(f'{col} vs churn')
    plt.legend(['Churn','No Churn'])
    plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(hue='churn', x=df_datos['account_charges_monthly'] * df_datos['customer_tenure'], data=df_datos, bins=30, kde=True)
plt.title('Total pagado vs Churn')
plt.legend(['Churn','No Churn'])
plt.show()
```
